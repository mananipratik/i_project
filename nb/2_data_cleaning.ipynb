### Invalid Value Check

```python
# ==========================================
# HR INVALID VALUE VALIDATION
# ==========================================

invalid_checks = {}

# Age should be positive and realistically within range
invalid_checks['Invalid Age'] = (
    (df1['Age'] < 18) |
    (df1['Age'] > 100)
).sum()


# Income should not be zero or negative
invalid_checks['Invalid MonthlyIncome'] = (
    df1['MonthlyIncome'] <= 0
).sum()


# Distance cannot be negative
invalid_checks['Invalid DistanceFromHome'] = (
    df1['DistanceFromHome'] < 0
).sum()


# Number of companies cannot be negative
invalid_checks['Invalid NumCompaniesWorked'] = (
    df1['NumCompaniesWorked'] < 0
).sum()


# Experience cannot be negative
experience_columns = [
    'TotalWorkingYears',
    'YearsAtCompany',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager'
]

for col in experience_columns:
    
    invalid_checks[f'Negative {col}'] = (
        df1[col] < 0
    ).sum()


# Training count cannot be negative
invalid_checks['Invalid TrainingTimesLastYear'] = (
    df1['TrainingTimesLastYear'] < 0
).sum()


invalid_value_report = pd.DataFrame(
    invalid_checks.items(),
    columns=[
        'Validation_Rule',
        'Invalid_Record_Count'
    ]
)

display(invalid_value_report)
```

### Categorical Consistency

```python
categorical_columns = df1.select_dtypes(
    include=['object', 'category']
).columns.tolist()

print("Categorical Columns:")
print(categorical_columns)
```

```python
# ==========================================
# CATEGORICAL CONSISTENCY ANALYSIS
# ==========================================

# 1. Identify categorical columns
categorical_columns = df1.select_dtypes(
    include=['object', 'category']
).columns.tolist()

print("Categorical Columns:")
print(categorical_columns)


# 2. Create consistency report
categorical_report = []

for col in categorical_columns:
    
    original_unique = df1[col].nunique(dropna=True)
    
    normalized_unique = (
        df1[col]
        .dropna()
        .astype(str)
        .str.strip()
        .str.lower()
        .nunique()
    )
    
    categorical_report.append({
        'Column': col,
        'Original_Unique': original_unique,
        'Normalized_Unique': normalized_unique,
        'Possible_Inconsistency':
            original_unique != normalized_unique
    })


categorical_consistency_report = pd.DataFrame(
    categorical_report
)

display(categorical_consistency_report)


# 3. Show detailed values only for
# potentially inconsistent columns
problem_columns = categorical_consistency_report[
    categorical_consistency_report[
        'Possible_Inconsistency'
    ]
]['Column'].tolist()

for col in problem_columns:
    
    print("\n" + "=" * 60)
    print(f"POSSIBLE INCONSISTENCY: {col}")
    print("=" * 60)
    
    print(df1[col].value_counts(dropna=False))
```

### Outlier Analysis

```python
numerical_columns = df1.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

print("Numerical Columns:")
print(numerical_columns)
```

```python
# ==========================================
# HR OUTLIER ANALYSIS
# IQR METHOD
# ==========================================

continuous_columns = [
    'Age',
    'DailyRate',
    'DistanceFromHome',
    'HourlyRate',
    'MonthlyIncome',
    'MonthlyRate',
    'NumCompaniesWorked',
    'PercentSalaryHike',
    'TotalWorkingYears',
    'TrainingTimesLastYear',
    'YearsAtCompany',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager'
]

# Keep only existing columns
continuous_columns = [
    col for col in continuous_columns
    if col in df1.columns
]


outlier_results = []

for col in continuous_columns:
    
    Q1 = df1[col].quantile(0.25)
    Q3 = df1[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)
    
    outlier_count = (
        (df1[col] < lower_bound) |
        (df1[col] > upper_bound)
    ).sum()
    
    outlier_percentage = (
        outlier_count / len(df1) * 100
    )
    
    outlier_results.append({
        'Column': col,
        'Lower_Bound': round(lower_bound, 2),
        'Upper_Bound': round(upper_bound, 2),
        'Outlier_Count': outlier_count,
        'Outlier_Percentage': round(
            outlier_percentage, 2
        )
    })


outlier_summary = pd.DataFrame(
    outlier_results
).sort_values(
    'Outlier_Count',
    ascending=False
)

display(outlier_summary)
```

### Cleaning Decisions

```python
# ==========================================
# DATA CLEANING
# ==========================================

# Create cleaned copy
df_clean = df.copy()

print("Original Shape:", df_clean.shape)


# ------------------------------------------
# 1. Remove exact duplicate rows
# ------------------------------------------

duplicates_before = df_clean.duplicated().sum()

if duplicates_before > 0:
    df_clean = df_clean.drop_duplicates()

duplicates_after = df_clean.duplicated().sum()

print("\nDuplicate Rows:")
print("Before:", duplicates_before)
print("After:", duplicates_after)


# ------------------------------------------
# 2. Missing values
# ------------------------------------------

print("\nMissing Values:")
print(df_clean.isnull().sum().sum())


# ------------------------------------------
# 3. No automatic outlier removal
# ------------------------------------------

print("\nOutliers:")
print("No rows removed based only on IQR.")


# ------------------------------------------
# 4. No automatic removal of
# logical inconsistency records
# ------------------------------------------

print("\nLogical Consistency:")
print(
    "Records flagged for investigation "
    "were retained."
)


# ------------------------------------------
# 5. Final shape
# ------------------------------------------

print("\nFinal Cleaned Shape:", df_clean.shape)
```

### Data Quality Report

```python
# ==========================================
# DAY 2 - INVALID VALUE CHECK
# ==========================================

invalid_results = []

# Age
invalid_age = (
    (df_clean['Age'] < 18) |
    (df_clean['Age'] > 100)
).sum()

invalid_results.append({
    'Validation_Rule': 'Age must be between 18 and 100',
    'Invalid_Record_Count': invalid_age
})


# Columns that cannot have negative values
non_negative_columns = [
    'DailyRate',
    'DistanceFromHome',
    'HourlyRate',
    'MonthlyIncome',
    'MonthlyRate',
    'NumCompaniesWorked',
    'PercentSalaryHike',
    'TotalWorkingYears',
    'TrainingTimesLastYear',
    'YearsAtCompany',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager'
]

for col in non_negative_columns:
    
    invalid_count = (df_clean[col] < 0).sum()
    
    invalid_results.append({
        'Validation_Rule': f'{col} must be >= 0',
        'Invalid_Record_Count': invalid_count
    })


invalid_value_report = pd.DataFrame(invalid_results)

display(invalid_value_report)
```

```python
# ==========================================
# ORDINAL VALUE VALIDATION
# ==========================================

ordinal_rules = {
    'Education': (1, 5),
    'EnvironmentSatisfaction': (1, 4),
    'JobInvolvement': (1, 4),
    'JobLevel': (1, 5),
    'JobSatisfaction': (1, 4),
    'PerformanceRating': (1, 4),
    'RelationshipSatisfaction': (1, 4),
    'StockOptionLevel': (0, 3),
    'WorkLifeBalance': (1, 4)
}

ordinal_results = []

for col, (min_val, max_val) in ordinal_rules.items():
    
    invalid_count = (
        (df_clean[col] < min_val) |
        (df_clean[col] > max_val)
    ).sum()
    
    ordinal_results.append({
        'Validation_Rule': f'{col} must be between {min_val} and {max_val}',
        'Invalid_Record_Count': invalid_count
    })


ordinal_validation_report = pd.DataFrame(ordinal_results)

display(ordinal_validation_report)
```

```python
final_invalid_value_report = pd.concat(
    [
        invalid_value_report,
        ordinal_validation_report
    ],
    ignore_index=True
)

display(final_invalid_value_report)
```
