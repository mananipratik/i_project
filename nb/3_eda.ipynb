### age analysis

```python
df['Age'].describe()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.hist(
    df['Age'],
    bins=10,
    edgecolor='black'
)

plt.title('Employee Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Employees')

plt.tight_layout()

plt.show()
```

```python
age_bins = [17, 25, 35, 45, 55, 65]

age_labels = [
    '18-25',
    '26-35',
    '36-45',
    '46-55',
    '56-65'
]

df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=age_bins,
    labels=age_labels
)

age_group_summary = (
    df['AgeGroup']
    .value_counts()
    .sort_index()
    .reset_index()
)

age_group_summary.columns = [
    'Age_Group',
    'Employee_Count'
]

age_group_summary['Percentage'] = (
    age_group_summary['Employee_Count']
    / age_group_summary['Employee_Count'].sum()
    * 100
).round(2)

display(age_group_summary)
```

### Career Variables Analysis

```python
career_columns = [
    'TotalWorkingYears',
    'YearsAtCompany',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager'
]

career_summary = pd.DataFrame({
    'Variable': career_columns,
    
    'Count': [
        df[col].count() for col in career_columns
    ],
    
    'Mean': [
        round(df[col].mean(), 2) for col in career_columns
    ],
    
    'Median': [
        df[col].median() for col in career_columns
    ],
    
    'Minimum': [
        df[col].min() for col in career_columns
    ],
    
    'Maximum': [
        df[col].max() for col in career_columns
    ],
    
    'Standard_Deviation': [
        round(df[col].std(), 2) for col in career_columns
    ]
})

display(career_summary)
```

```python
reliable_career_columns = [
    'TotalWorkingYears',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager'
]

for col in reliable_career_columns:
    
    plt.figure(figsize=(10, 6))
    
    plt.hist(
        df[col],
        bins=10,
        edgecolor='black'
    )
    
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Number of Employees')
    
    plt.tight_layout()

    plt.show()
```

```python
experience_bins = [-1, 5, 10, 20, float('inf')]

experience_labels = [
    '0-5 Years',
    '6-10 Years',
    '11-20 Years',
    '20+ Years'
]

df['ExperienceGroup'] = pd.cut(
    df['TotalWorkingYears'],
    bins=experience_bins,
    labels=experience_labels
)

experience_summary = (
    df['ExperienceGroup']
    .value_counts()
    .sort_index()
    .reset_index()
)

experience_summary.columns = [
    'Experience_Group',
    'Employee_Count'
]

experience_summary['Percentage'] = (
    experience_summary['Employee_Count']
    / experience_summary['Employee_Count'].sum()
    * 100
).round(2)

display(experience_summary)
```

```python
promotion_bins = [-1, 1, 3, 5, float('inf')]

promotion_labels = [
    '0-1 Years',
    '2-3 Years',
    '4-5 Years',
    '6+ Years'
]

df['PromotionGapGroup'] = pd.cut(
    df['YearsSinceLastPromotion'],
    bins=promotion_bins,
    labels=promotion_labels
)

promotion_summary = (
    df['PromotionGapGroup']
    .value_counts()
    .sort_index()
    .reset_index()
)

promotion_summary.columns = [
    'Promotion_Gap_Group',
    'Employee_Count'
]

promotion_summary['Percentage'] = (
    promotion_summary['Employee_Count']
    / promotion_summary['Employee_Count'].sum()
    * 100
).round(2)

display(promotion_summary)
```

```python
manager_bins = [-1, 2, 5, 10, float('inf')]

manager_labels = [
    '0-2 Years',
    '3-5 Years',
    '6-10 Years',
    '10+ Years'
]

df['ManagerDurationGroup'] = pd.cut(
    df['YearsWithCurrManager'],
    bins=manager_bins,
    labels=manager_labels
)

manager_duration_summary = (
    df['ManagerDurationGroup']
    .value_counts()
    .sort_index()
    .reset_index()
)

manager_duration_summary.columns = [
    'Manager_Duration_Group',
    'Employee_Count'
]

manager_duration_summary['Percentage'] = (
    manager_duration_summary['Employee_Count']
    / manager_duration_summary['Employee_Count'].sum()
    * 100
).round(2)

display(manager_duration_summary)
```

### Training Analysis

### Attrition analysis

```python
total_employees = len(df)

employees_left = df['Attrition'].sum()

employees_stayed = total_employees - employees_left

attrition_rate = round(
    employees_left / total_employees * 100,
    2
)

overall_attrition_summary = pd.DataFrame({
    'Metric': [
        'Total Employees',
        'Employees Stayed',
        'Employees Left',
        'Attrition Rate (%)'
    ],

    'Value': [
        total_employees,
        employees_stayed,
        employees_left,
        attrition_rate
    ]
})

display(overall_attrition_summary)
```

```python
attrition_counts = (
    df['Attrition']
    .value_counts()
    .sort_index()
)

attrition_labels = ['Stayed', 'Left']

plt.figure(figsize=(7, 5))

plt.bar(
    attrition_labels,
    attrition_counts.values,
    edgecolor='black'
)

plt.title('Overall Employee Attrition')
plt.xlabel('Employee Status')
plt.ylabel('Number of Employees')

plt.tight_layout()

plt.show()
```

```python

department_attrition = (
    df.groupby('Department')
    .agg(
        Total_Employees=('Attrition', 'count'),
        Employees_Left=('Attrition', 'sum')
    )
    .reset_index()
)

department_attrition['Attrition_Rate (%)'] = (
    department_attrition['Employees_Left']
    / department_attrition['Total_Employees']
    * 100
).round(2)

department_attrition = department_attrition.sort_values(
    by='Attrition_Rate (%)',
    ascending=False
)

display(department_attrition)
```

```python
department_chart = department_attrition.sort_values(
    by='Attrition_Rate (%)',
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    department_chart['Department'],
    department_chart['Attrition_Rate (%)'],
    edgecolor='black'
)

plt.title('Attrition Rate by Department')
plt.xlabel('Attrition Rate (%)')
plt.ylabel('Department')

plt.tight_layout()

plt.show()
```

```python

job_role_attrition = (
    df.groupby('JobRole')
    .agg(
        Total_Employees=('Attrition', 'count'),
        Employees_Left=('Attrition', 'sum')
    )
    .reset_index()
)

job_role_attrition['Attrition_Rate (%)'] = (
    job_role_attrition['Employees_Left']
    / job_role_attrition['Total_Employees']
    * 100
).round(2)

job_role_attrition = job_role_attrition.sort_values(
    by='Attrition_Rate (%)',
    ascending=False
)

display(job_role_attrition)
```

```python
job_role_chart = job_role_attrition.sort_values(
    by='Attrition_Rate (%)',
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    job_role_chart['JobRole'],
    job_role_chart['Attrition_Rate (%)'],
    edgecolor='black'
)

plt.title('Attrition Rate by Job Role')
plt.xlabel('Attrition Rate (%)')
plt.ylabel('Job Role')

plt.tight_layout()

plt.show()
```

```python
job_level_attrition = (
    df.groupby('JobLevel')
    .agg(
        Total_Employees=('Attrition', 'count'),
        Employees_Left=('Attrition', 'sum')
    )
    .reset_index()
)

job_level_attrition['Attrition_Rate (%)'] = (
    job_level_attrition['Employees_Left']
    / job_level_attrition['Total_Employees']
    * 100
).round(2)

job_level_attrition = job_level_attrition.sort_values(
    by='JobLevel'
)

display(job_level_attrition)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    job_level_attrition['JobLevel'].astype(str),
    job_level_attrition['Attrition_Rate (%)'],
    edgecolor='black'
)

plt.title('Attrition Rate by Job Level')
plt.xlabel('Job Level')
plt.ylabel('Attrition Rate (%)')

plt.tight_layout()

plt.show()
```

```python
highest_attrition_job_level = job_level_attrition.loc[
    job_level_attrition['Attrition_Rate (%)'].idxmax()
]

print(highest_attrition_job_level)
```

```python

attrition_analysis_report = pd.concat([
    department_attrition,
    job_role_attrition,
    job_level_attrition
], ignore_index=True)

display(attrition_analysis_report)
```
